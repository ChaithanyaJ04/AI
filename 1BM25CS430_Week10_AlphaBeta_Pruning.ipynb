```python
def leaves(tree):
    if isinstance(tree, int):
        return [tree]
    return leaves(tree[0]) + leaves(tree[1])

def alpha_beta(tree, depth, alpha, beta, path):
    if isinstance(tree, int):
        print(path, "=", tree)
        return tree

    if depth % 2 == 0:
        best = float('-inf')
        node = "MAX"
    else:
        best = float('inf')
        node = "MIN"

    for i in range(2):
        side = "L" if i == 0 else "R"
        child_path = path + "-" + side

        value = alpha_beta(tree[i], depth + 1,
                           alpha, beta, child_path)

        if node == "MAX":
            best = max(best, value)
            alpha = max(alpha, best)
        else:
            best = min(best, value)
            beta = min(beta, best)

        if alpha >= beta:
            for j in range(i + 1, 2):
                p = path + ("-L" if j == 0 else "-R")
                print("Pruned:", p, "=", leaves(tree[j]))
            break

    print(node, path, "=", best,
          "Alpha =", alpha, "Beta =", beta)

    return best


tree = [
    [[10, 9], [14, 18]],
    [[5, 4], [50, 3]]
]

print("Alpha-Beta Pruning Steps\n")

result = alpha_beta(tree, 0, float('-inf'),
                    float('inf'), "Root")

print("\nFinal MAX value =", result)
print("Optimal path = Root -> L -> L -> 10")
```
