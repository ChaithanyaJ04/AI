```python
john = input("Enter person's name who likes food: ")
anil = input("Enter person's name who eats peanuts: ")
harry = input("Enter another person's name: ")
food = input("Enter food item: ")

print("\n--- Knowledge Base ---")
print("Food(Apple)")
print("Food(Vegetables)")
print("Eats(" + anil + "," + food + ")")
print("Alive(" + anil + ")")

print("\n--- FOL Representation ---")
print("1. Food(x) -> Likes(" + john + ",x)")
print("2. Eats(x,y) AND NOT Killed(x) -> Food(y)")
print("3. Eats(" + anil + ",y) -> Eats(" + harry + ",y)")
print("4. Alive(x) -> NOT Killed(x)")
print("5. NOT Killed(x) -> Alive(x)")

print("\n--- CNF Clauses ---")
print("C1: NOT Food(x) OR Likes(" + john + ",x)")
print("C2: NOT Eats(x,y) OR Killed(x) OR Food(y)")
print("C3: NOT Eats(" + anil + ",y) OR Eats(" + harry + ",y)")
print("C4: NOT Alive(x) OR NOT Killed(x)")
print("C5: Killed(x) OR Alive(x)")

print("\n--- Negated Goal ---")
print("NOT Likes(" + john + "," + food + ")")

print("\n--- Resolution Steps ---")

print("\nStep 1:")
print("Alive(" + anil + ")")
print("NOT Alive(x) OR NOT Killed(x)")
print("=> NOT Killed(" + anil + ")")

print("\nStep 2:")
print("NOT Killed(" + anil + ")")
print("NOT Eats(x,y) OR Killed(x) OR Food(y)")
print("=> NOT Eats(" + anil + "," + food +
      ") OR Food(" + food + ")")

print("\nStep 3:")
print("Eats(" + anil + "," + food + ")")
print("=> Food(" + food + ")")

print("\nStep 4:")
print("Food(" + food + ")")
print("NOT Food(x) OR Likes(" + john + ",x)")
print("=> Likes(" + john + "," + food + ")")

print("\nStep 5:")
print("Likes(" + john + "," + food + ")")
print("NOT Likes(" + john + "," + food + ")")
print("=> EMPTY CLAUSE")

print("\n--- Resolution Proof Tree ---")
print("             Alive(" + anil + ")")
print("                    |")
print("          NOT Killed(" + anil + ")")
print("                    |")
print("    NOT Eats(" + anil + "," + food + ")")
print("          OR Food(" + food + ")")
print("                    |")
print("             Food(" + food + ")")
print("                    |")
print("       Likes(" + john + "," + food + ")")
print("                    |")
print("   NOT Likes(" + john + "," + food + ")")
print("                    |")
print("              EMPTY CLAUSE")

print("\n--- Final Result ---")
print("Goal: Likes(" + john + "," + food + ")")
print(john + " likes " + food)
print("Query Proved Successfully!")
```
