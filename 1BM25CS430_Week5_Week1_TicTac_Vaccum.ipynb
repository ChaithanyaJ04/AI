```python
#TIC TAC TOE
board = [" "] * 9

def print_board():
    print(f"\n {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} \n")

def check_win(p):
    lines = [(0,1,2), (3,4,5), (6,7,8), (0,3,6), (1,4,7), (2,5,8), (0,4,8), (2,4,6)]
    return any(board[a] == board[b] == board[c] == p for a, b, c in lines)

player = "X"

for turn in range(9):
    print_board()

    if player == "X":
        pos = int(input("Enter position (0-8): "))
        board[pos] = "X"
    else:
        pos = board.index(" ")  # Computer picks first empty spot
        print(f"Computer (O) places at position: {pos}")
        board[pos] = "O"

    if check_win(player):
        print_board()
        print(f"Player {player} Wins!")
        break

    player = "O" if player == "X" else "X"
else:
    print_board()
    print("It's a Draw!")
```

```python
def reflex_vacuum_agent(location, status_a, status_b):
    if location == "A":
        if status_a == "DIRTY":
            return "SUCK"
        elif status_b == "DIRTY":
            return "MOVE_RIGHT"
        else:
            return "NO_OP"
    elif location == "B":
        if status_b == "DIRTY":
            return "SUCK"
        elif status_a == "DIRTY":
            return "MOVE_LEFT"
        else:
            return "NO_OP"

def run_vacuum_cleaner():
    print("Vacuum Cleaner Agent Simulation!")

    location = input("Enter initial agent location (A/B): ").strip().upper()
    status_a = input("Enter status of Room A (DIRTY/CLEAN): ").strip().upper()
    status_b = input("Enter status of Room B (DIRTY/CLEAN): ").strip().upper()

    print("\nStarting Environment")
    print(f"Location: {location} | Room A: {status_a} | Room B: {status_b}\n")

    step = 1
    while True:
        action = reflex_vacuum_agent(location, status_a, status_b)
        print(f"Step {step}: Agent in Room {location} | Action -> {action}")

        if action == "SUCK":
            if location == "A":
                status_a = "CLEAN"
            else:
                status_b = "CLEAN"
            print(f"  -> Cleaned Room {location}. State now: (Room A: {status_a}, Room B: {status_b})")

        elif action == "MOVE_RIGHT":
            location = "B"
            print(f"  -> Moved to Room B.")

        elif action == "MOVE_LEFT":
            location = "A"
            print(f"  -> Moved to Room A.")

        elif action == "NO_OP":
            print("\nGoal Reached: All rooms are clean!")
            break

        step += 1

if __name__ == "__main__":
    run_vacuum_cleaner()
```
