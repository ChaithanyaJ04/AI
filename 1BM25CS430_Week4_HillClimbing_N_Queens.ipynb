```python
def calculate_cost(state):
    n = len(state)
    cost = 0

    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def hill_climbing(n=4):
    state = [3, 1, 2, 0]

    while True:
        current_cost = calculate_cost(state)

        print("Current state:", state)
        print("Current cost:", current_cost)

        if current_cost == 0:
            print("Goal configuration reached!")
            return state

        best_state = state[:]
        best_cost = current_cost

        for col in range(n):
            for row in range(1, n + 1):
                if row != state[col]:
                    neighbor = state[:]
                    neighbor[col] = row
                    neighbor_cost = calculate_cost(neighbor)

                    print("Neighbor:", neighbor,
                          "Cost:", neighbor_cost)

                    if neighbor_cost < best_cost:
                        best_cost = neighbor_cost
                        best_state = neighbor

        if best_cost >= current_cost:
            print("Local optimum reached. Restart needed.")
            return None

        state = best_state


solution = hill_climbing()

if solution:
    print("Solution:", solution)
```

```python
import random
import math

def calculate_cost(state):
    cost = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if (state[i] == state[j] or
                abs(state[i] - state[j]) == abs(i - j)):
                cost += 1

    return cost


def simulated_annealing(n=8):
    random.seed(42)

    current = [random.randint(1, n) for _ in range(n)]
    current_cost = calculate_cost(current)

    T = 10.0
    cooling_rate = 0.995
    step = 0

    print("Initial State:", current)
    print("Initial Cost:", current_cost)

    while T > 0.01 and current_cost != 0:
        next_state = current[:]

        col = random.randrange(n)
        row = random.randint(1, n)

        while row == next_state[col]:
            row = random.randint(1, n)

        next_state[col] = row
        next_cost = calculate_cost(next_state)

        delta_E = current_cost - next_cost

        if delta_E > 0:
            current = next_state
            current_cost = next_cost
            status = "Accepted"
        else:
            probability = math.exp(delta_E / T)

            if random.random() < probability:
                current = next_state
                current_cost = next_cost
                status = "Accepted"
            else:
                status = "Rejected"

        step += 1

        if step <= 5 or step % 100 == 0 or current_cost == 0:
            print("\nStep:", step)
            print("Next State:", next_state)
            print("Next Cost:", next_cost)
            print("Delta E:", delta_E)
            print("Temperature:", round(T, 3))
            print("Status:", status)
            print("Current State:", current)
            print("Current Cost:", current_cost)

        T *= cooling_rate

    if current_cost == 0:
        print("\nGoal Configuration Found!")
        print("Final State:", current)
        print("Final Cost:", current_cost)

        print("\nChessboard:")
        for row in range(1, n + 1):
            for col in range(n):
                if current[col] == row:
                    print("Q", end=" ")
                else:
                    print("-", end=" ")
            print()
    else:
        print("\nSolution not found in this run.")


simulated_annealing()

```
