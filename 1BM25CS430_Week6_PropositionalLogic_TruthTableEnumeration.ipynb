```python
from itertools import product

# Function to check whether a sentence is true in a model
def pl_true(sentence, model):
    return eval(sentence, {"__builtins__": {}}, model)


# Truth Table Enumeration Algorithm
def tt_entails(KB, query):

    # Get all propositional symbols
    symbols = ['P', 'Q', 'R']

    # Generate all possible truth assignments
    for values in product([True, False], repeat=len(symbols)):

        model = dict(zip(symbols, values))

        # Check whether KB is true
        kb_true = all(pl_true(sentence, model) for sentence in KB)

        if kb_true:
            # If KB is true but query is false
            if not pl_true(query, model):
                return False

    return True


# Knowledge Base
KB = [
    "(not Q) or P",       # Q -> P
    "(not P) or (not Q)", # P -> not Q
    "Q or R"              # Q or R
]


# Display truth table
print("P\tQ\tR\tKB")

for values in product([True, False], repeat=3):

    model = dict(zip(['P', 'Q', 'R'], values))

    result = all(pl_true(sentence, model) for sentence in KB)

    print(values[0], values[1], values[2], result, sep="\t")


# Queries
queries = {
    "R": "R",
    "R -> P": "(not R) or P",
    "Q -> R": "(not Q) or R"
}

# Check entailment
for name, query in queries.items():

    result = tt_entails(KB, query)

    print("\nDoes KB entail", name, "?")

    if result:
        print("YES")
    else:
        print("NO")
```
